```python
df = pd.read_csv("animal-fun-facts-dataset.csv")
print("Loaded rows:", len(df))
df.head()
```

```python
df.text
```

```python
for i, row in df.iterrows():
    print(f"Animal name: {row['animal_name']}, Fact: {row['text']}")
    # print(f"Modified: {row['text2']}")
```

```python
def ensure_animal_in_text(row):
    name = str(row["animal_name"]).strip()
    text = str(row["text"]).strip()

    if name.lower() in text.lower():
        return text

    name_cap = name.capitalize()

    # find first alphabetic character to lowercase safely
    match = re.search(r"[A-Za-z]", text)
    if match:
        i = match.start()
        text = text[:i] + text[i].lower() + text[i+1:]

    return f"{name_cap}, {text}"

df["animal_fact"] = df.apply(ensure_animal_in_text, axis=1)

df["animal_fact"]
```

```python
df = df[["animal_name", "animal_fact"]]
df.head()
```

```python
df.to_csv("AnimalFacts.csv", index=False)
```
